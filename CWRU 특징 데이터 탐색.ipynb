```python
# ============================================================
# Step 1: CWRU 특징 데이터 탐색
# 목적: 학습 전에 "내 데이터가 어떻게 생겼는지" 눈으로 확인
# ============================================================
import pandas as pd

# CSV 불러오기 (경로는 네 환경에 맞게 수정)
# 탭(tab)으로 구분된 파일이므로, pd.read_table을 사용합니다.
df = pd.read_table('cwru_features.csv')

print("=" * 50)
print(f"전체 데이터: {df.shape[0]}행, {df.shape[1]}열")
print("=" * 50)

# 1) 고장 유형별로 몇 개씩 있나? (클래스 균형 확인)
print("\n[고장 유형별 개수]")
print(df['fault_type'].value_counts())

# 2) 카테고리(측정 조건)별 분포
print("\n[측정 카테고리별 개수]")
print(df['category'].value_counts())

# 3) 컬럼(특징) 이름들 살펴보기
print("\n[사용 가능한 특징 컬럼들]")
print([c for c in df.columns if c.startswith('DE_')][:15], "...")

# 4) 결측치(빈 칸) 있는지 확인 — 아까 B028/IR028은 FE가 비어있었음
print("\n[결측치가 있는 컬럼 개수]")
print(df.isnull().sum()[df.isnull().sum() > 0])
```
