```python
# ============================================================
# 🔥 AI 김반장 Phase A — CWRU 베어링 4종 분류 (풀 파이프라인)
# 데이터로드 → 전처리 → 1D-CNN 학습 → 평가 → 혼동행렬 시각화
# ============================================================

# ---- [0] 준비: 라이브러리 & 파일 업로드 ----
import os
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, f1_score
import matplotlib.pyplot as plt

# 파일 없으면 업로드 창 띄우기 (Colab 전용)
CSV = 'cwru_features.csv'
if not os.path.exists(CSV):
    try:
        from google.colab import files
        print("📂 cwru_features.csv 파일을 선택해 업로드해줘...")
        up = files.upload()
        CSV = list(up.keys())[0]
    except Exception:
        raise FileNotFoundError("cwru_features.csv를 같은 폴더에 두거나 Colab에 업로드해줘")

np.random.seed(42)
tf.random.set_seed(42)   # 재현성 고정 (시뮬레이터 인증에서 중요했던 부분!)

# ---- [1] 데이터 로드 & 살펴보기 ----
# ---- [1] 데이터 로드 (구분자 자동 감지로 수정!) ----
# sep=None + engine='python' → 콤마든 탭이든 알아서 판단
df = pd.read_csv(CSV, sep=None, engine='python')

print("=" * 55)
print(f"전체 데이터: {df.shape[0]}행 x {df.shape[1]}열")

# ★ 검증: 컬럼이 제대로 쪼개졌는지 반드시 확인!
print(f"\n읽어들인 컬럼 개수: {len(df.columns)}")
if len(df.columns) < 5:
    print("⚠️ 컬럼이 안 쪼개졌어! 구분자 문제일 수 있음")
    print("첫 컬럼 이름:", df.columns[0][:80])
else:
    print("✅ 컬럼 정상 분리됨")

print("\n[고장 유형별 개수]")
print(df['fault_type'].value_counts())
print("=" * 55)

# ---- [2] 특징(X)과 정답(y) 분리 ----
# DE_ 로 시작하는 진동 특징 컬럼만 사용 (Drive End)
feature_cols = [c for c in df.columns if c.startswith('DE_')
                and df[c].dtype != object]
X = df[feature_cols].copy()
y_raw = df['fault_type'].copy()

# 결측치(028 계열 등) → 각 컬럼 평균으로 채우기
X = X.fillna(X.mean())
print(f"\n사용 특징 개수: {X.shape[1]}개")

# 정답 라벨을 숫자로 변환 (Normal=0, IR=1, Ball=2, OR=3 식)
classes = sorted(y_raw.unique())
class_to_id = {c: i for i, c in enumerate(classes)}
id_to_class = {i: c for c, i in class_to_id.items()}
y = y_raw.map(class_to_id).values
print(f"클래스 매핑: {class_to_id}")

# ---- [3] 정규화 + train/test 분리 ----
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.25, random_state=42, stratify=y)

# 1D-CNN 입력 형태로 reshape: (샘플수, 특징수, 1)
X_train = X_train[..., np.newaxis]
X_test = X_test[..., np.newaxis]
print(f"\n학습 데이터: {X_train.shape}, 테스트: {X_test.shape}")

# ---- [4] 1D-CNN 모델 설계 ----
num_classes = len(classes)
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_train.shape[1], 1)),
    tf.keras.layers.Conv1D(32, kernel_size=3, padding='same', activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Conv1D(64, kernel_size=3, padding='same', activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(num_classes, activation='softmax')
])
model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
model.summary()

# ---- [5] 학습! ----
print("\n🚀 학습 시작...")
history = model.fit(X_train, y_train,
                    validation_split=0.2,
                    epochs=60, batch_size=16, verbose=0)
print("✅ 학습 완료!")

# ---- [6] 평가 ----
loss, acc = model.evaluate(X_test, y_test, verbose=0)
y_pred = model.predict(X_test, verbose=0).argmax(axis=1)
macro_f1 = f1_score(y_test, y_pred, average='macro')

print("\n" + "=" * 55)
print(f"🎯 테스트 정확도(Accuracy) : {acc*100:.2f}%")
print(f"🎯 Macro F1 Score          : {macro_f1:.4f}")
print("=" * 55)
print("\n[클래스별 상세 성적표]")
print(classification_report(y_test, y_pred,
      target_names=[id_to_class[i] for i in range(num_classes)]))

# ---- [7] 시각화: 학습곡선 + 혼동행렬 ----
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 학습곡선
ax1.plot(history.history['accuracy'], label='train')
ax1.plot(history.history['val_accuracy'], label='validation')
ax1.set_title('Learning Curve')
ax1.set_xlabel('Epoch'); ax1.set_ylabel('Accuracy')
ax1.legend(); ax1.grid(alpha=0.3)

# 혼동행렬
cm = confusion_matrix(y_test, y_pred)
im = ax2.imshow(cm, cmap='Blues')
ax2.set_title('Confusion Matrix')
ax2.set_xlabel('Predicted'); ax2.set_ylabel('True')
ax2.set_xticks(range(num_classes)); ax2.set_yticks(range(num_classes))
ax2.set_xticklabels([id_to_class[i] for i in range(num_classes)])
ax2.set_yticklabels([id_to_class[i] for i in range(num_classes)])
for i in range(num_classes):
    for j in range(num_classes):
        ax2.text(j, i, cm[i, j], ha='center', va='center',
                 color='white' if cm[i, j] > cm.max()/2 else 'black')
plt.colorbar(im, ax=ax2)
plt.tight_layout()
plt.show()

print("\n🎉 축하해 HoonNi! AI가 베어링 고장을 구분하기 시작했어!")
```
