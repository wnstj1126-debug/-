```python
"""
김반장(Deep_Bind) - 상용전원(DOL) 모터 MCSA 정상상태 FFT 검증 시뮬레이션
목적: 회전자 바 결함(broken rotor bar) 사이드밴드를 가상 생성하고 FFT로 검출

작성 의도:
  1) 정상 모터 vs 결함 모터의 전류 신호를 직접 만들어 본다
  2) FFT + 윈도우로 -40~-60dB 사이드밴드를 잡아내는 과정을 검증한다
  3) 캡처 시간(주파수 분해능)과 윈도우 선택의 중요성을 체감한다
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import get_window

# ----------------------------------------------------------------------
# 1. 시뮬레이션 파라미터 (실제 김반장 ADC 설정과 대응)
# ----------------------------------------------------------------------
fs = 5000          # 샘플링 주파수 [Hz]  (실제 STM32 ADC 샘플레이트에 대응)
capture_sec = 12    # 캡처 시간 [초]  -> 주파수 분해능 = 1/12 ≈ 0.083Hz
                   #    사이드밴드를 기본파와 분리하려면 분해능이 충분히 작아야 함

f_line = 60.0      # 라인 주파수 [Hz]  (한국은 60Hz)
slip   = 0.03      # 슬립 s  (부하 걸린 유도전동기 전형값 0.01~0.05)

# 결함 심각도: 사이드밴드 크기를 기본파 대비 dB로 지정
#   정상 모터: 매우 작음(-60dB 이하) / 결함 모터: 커짐(-40dB 부근)
sideband_db = -45.0   # <-- 이 값을 -60(정상) ~ -35(심한 결함)으로 바꿔보며 실험!

noise_db = -70.0   # 배경 노이즈 레벨 [dB] (기본파 대비)

# ----------------------------------------------------------------------
# 2. 가상 전류 신호 생성
# ----------------------------------------------------------------------
N = int(fs * capture_sec)          # 총 샘플 수
t = np.arange(N) / fs              # 시간 축

# (a) 기본파 - 진폭 1.0 기준
A_fund = 1.0
current = A_fund * np.sin(2 * np.pi * f_line * t)

# (b) 회전자 바 결함 사이드밴드: f_line * (1 ± 2*k*s), k=1,2
#     dB -> 선형 진폭 변환:  amp = 10^(dB/20)
A_side = A_fund * 10 ** (sideband_db / 20.0)

for k in [1, 2]:                   # 1차, 2차 사이드밴드까지 생성
    f_lower = f_line * (1 - 2 * k * slip)
    f_upper = f_line * (1 + 2 * k * slip)
    # k가 커질수록 진폭은 작아지는 게 일반적 -> k로 한 번 더 감쇠
    amp_k = A_side / k
    current += amp_k * np.sin(2 * np.pi * f_lower * t)
    current += amp_k * np.sin(2 * np.pi * f_upper * t)

# (c) 약간의 하모닉 (실제 전원에는 3,5,7고조파가 늘 있음) - 검출 방해 요소
for h, h_db in [(3, -30), (5, -34), (7, -40)]:
    current += A_fund * 10 ** (h_db / 20.0) * np.sin(2 * np.pi * f_line * h * t)

# (d) 백색 노이즈
noise_amp = A_fund * 10 ** (noise_db / 20.0)
current += noise_amp * np.random.randn(N)

# ----------------------------------------------------------------------
# 3. FFT 분석 (윈도우 적용이 핵심!)
# ----------------------------------------------------------------------
# 윈도우: 기본파의 스펙트럼 누설(leakage)을 억제해야 옆의 작은 사이드밴드가 보임
#   - 'hann'도 괜찮지만, 사이드로브가 더 낮은 'blackmanharris'가 MCSA에 유리
win = get_window('blackmanharris', N)
win_gain = np.sum(win) / N         # 윈도우 진폭 보정 계수

xw = current * win
spectrum = np.fft.rfft(xw)
freqs = np.fft.rfftfreq(N, d=1/fs)

# 진폭 정규화 후 dB 변환 (기본파를 0dB 기준으로)
mag = np.abs(spectrum) / N * 2 / win_gain
mag_db = 20 * np.log10(mag / np.max(mag) + 1e-12)   # 최대값(기본파) 기준 0dB

# ----------------------------------------------------------------------
# 4. 사이드밴드 검출 (결함 지표 계산)
# ----------------------------------------------------------------------
def peak_db_near(target_freq, search_hz=0.5):
    """target_freq 근처에서 가장 큰 피크의 dB값과 실제 주파수를 반환"""
    mask = (freqs > target_freq - search_hz) & (freqs < target_freq + search_hz)
    if not np.any(mask):
        return None, None
    idx = np.where(mask)[0]
    peak_idx = idx[np.argmax(mag_db[idx])]
    return freqs[peak_idx], mag_db[peak_idx]

print("=" * 60)
print(f"라인주파수 {f_line}Hz, 슬립 {slip}, 캡처 {capture_sec}s")
print(f"주파수 분해능 = {fs/N:.4f} Hz")
print("=" * 60)

f_fund, db_fund = peak_db_near(f_line)
print(f"[기본파]   {f_fund:.2f} Hz : {db_fund:6.2f} dB")

# 1차 사이드밴드 검출
f_l = f_line * (1 - 2 * slip)
f_u = f_line * (1 + 2 * slip)
fl_det, dbl = peak_db_near(f_l)
fu_det, dbu = peak_db_near(f_u)
print(f"[하측 SB]  {fl_det:.2f} Hz : {dbl:6.2f} dB  (이론값 {f_l:.2f}Hz)")
print(f"[상측 SB]  {fu_det:.2f} Hz : {dbu:6.2f} dB  (이론값 {f_u:.2f}Hz)")

# 결함 지표: 사이드밴드 평균 dB (기본파=0dB 기준)
fault_index = (dbl + dbu) / 2
print("-" * 60)
print(f"[결함지표] 사이드밴드 평균 = {fault_index:.2f} dB (기본파 대비)")

# 간단한 판정 임계값 (현장 데이터로 튜닝해야 할 값)
THRESHOLD_DB = -50.0
if fault_index > THRESHOLD_DB:
    print(f"  ==> 경고: 회전자 바 결함 의심 (임계 {THRESHOLD_DB}dB 초과)")
else:
    print(f"  ==> 정상 범위 (임계 {THRESHOLD_DB}dB 이하)")
print("=" * 60)

# ----------------------------------------------------------------------
# 5. 시각화
# ----------------------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 8))

# 전체 스펙트럼
ax1.plot(freqs, mag_db, lw=0.8)
ax1.set_xlim(0, 500)
ax1.set_ylim(-100, 5)
ax1.set_xlabel('Frequency [Hz]')
ax1.set_ylabel('Magnitude [dB]')
ax1.set_title('MCSA Full Spectrum (DOL Motor)')
ax1.grid(True, alpha=0.3)

# 기본파 주변 확대 (사이드밴드가 보이는 핵심 영역)
ax2.plot(freqs, mag_db, lw=1.0)
ax2.axvline(f_l, color='r', ls='--', alpha=0.6, label=f'Lower SB {f_l:.1f}Hz')
ax2.axvline(f_u, color='g', ls='--', alpha=0.6, label=f'Upper SB {f_u:.1f}Hz')
ax2.set_xlim(f_line - 12, f_line + 12)
ax2.set_ylim(-100, 5)
ax2.set_xlabel('Frequency [Hz]')
ax2.set_ylabel('Magnitude [dB]')
ax2.set_title('Zoom around Line Frequency (sidebands visible)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
